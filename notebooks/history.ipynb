```python
import pandas as pd
import pathlib
```

```python
project_root = pathlib.Path.cwd().resolve()
if project_root.name == "notebooks":
    project_root = project_root.parent

fixtures_dir = project_root / "data" / "history" / "fixtures"
gws_dir = project_root / "data" / "history" / "gws"
```

```python
fixtures_path = pathlib.Path(fixtures_dir)

csv_files = fixtures_path.glob("*.csv")

# fixtures_list = [pd.read_csv(file) for file in csv_files]

fixtures_list = []

for file in csv_files:
    df = pd.read_csv(file)

    season_code = file.stem.split("-")[-1]
    season = f"{season_code[:2]}/{season_code[2:]}"
    df["season"] = season

    fixtures_list.append(df)

combined_fixtures = pd.concat(fixtures_list, ignore_index=True)
```

```python
gws_path = pathlib.Path(gws_dir)

csv_files = gws_path.glob("*.csv")

# gws_list = [pd.read_csv(file) for file in csv_files]

gws_list = []

for file in csv_files:
    try:
        df = pd.read_csv(file, encoding="utf-8")

        season_code = file.stem.split("-")[-1]
        season = f"{season_code[:2]}/{season_code[2:]}"
        df["season"] = season


        gws_list.append(df)
        print( f"Succes on read: {file}")
    except UnicodeDecodeError:
        df = pd.read_csv(file, encoding="latin-1")

        season_code = file.stem.split("-")[-1]
        season = f"{season_code[:2]}/{season_code[2:]}"
        df["season"] = season

        gws_list.append(df)
        print( f"Succes on read: {file}")


combined_gws = pd.concat(gws_list, ignore_index=True)
```

```python
sliced_df = combined_gws[[
    'name', 'position', 'team', 'xP', 'total_points', 'expected_assists',
    'expected_goal_involvements', 'expected_goals',
    'fixture', 'goals_conceded', 'goals_scored','opponent_team',
    'round', 'saves', 'was_home', 'GW', 'season'
]]

sliced_df = sliced_df[sliced_df["position"] != "AM"]

sliced_df.sort_values('xP', ascending=False)[:15]
```
